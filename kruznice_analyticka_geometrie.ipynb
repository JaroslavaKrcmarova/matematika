
# 📐 Analytická geometrie v rovině – Kružnice



## 🔹 Motivace

Kružnice se v matematice neobjevuje náhodou. Setkáme se s ní například:

- při návrhu **kol, ozubených převodů a ložisek**
- při určování **dosahu signálu** (Wi‑Fi, mobilní sítě)
- v **kartografii** – oblast stejně vzdálená od určitého bodu
- při pohybu těles, která se **otáčejí kolem pevného středu**

Ve všech případech jde o množinu bodů, které mají **stejnou vzdálenost od jednoho bodu**.



## 🔹 Definice a základní vlastnosti

**Kružnice** je množina všech bodů v rovině, které mají od pevného bodu  
$S$ (střed kružnice) stejnou vzdálenost $r$ (poloměr).

- střed kružnice: $S=[m,n]$
- poloměr kružnice: $r>0$



## 🔹 Středový tvar rovnice kružnice

Rovnice kružnice se středem $S=[m,n]$ a poloměrem $r$ má tvar:

$$
(x-m)^2 + (y-n)^2 = r^2
$$

### Z rovnice ihned určíme:
- střed $S=[m,n]$
- poloměr $r=\sqrt{r^2}$



### 📌 Příklad

$$
(x+2)^2 + (y-5)^2 = 5
$$

- $S=[-2,5]$
- $r=\sqrt{5}$



## 🔹 Obecná rovnice kružnice

Obecná rovnice kružnice má tvar:

$$
x^2 + y^2 + ax + by + c = 0
$$

Z této rovnice **nelze přímo odečíst střed a poloměr**, proto ji musíme převést  
na středový tvar pomocí **doplnění na čtverec**.



## 🔹 Úprava obecné rovnice na středový tvar  
### (doplnění na čtverec)

$$
x^2 + y^2 + 6x + 4y - 48 = 0
$$

Seskupíme členy:

$$
(x^2 + 6x) + (y^2 + 4y) = 48
$$

Doplníme na čtverce:

$$
(x+3)^2 + (y+2)^2 = 61
$$

- střed: $S=[-3,-2]$
- poloměr: $r=\sqrt{61}$



## 🔹 Ukázka řešených příkladů

### ✏️ Příklad 1
Urči střed a poloměr kružnice:

$$
(x-5)^2 + (y+3)^2 = 9
$$

**Řešení:**  
$S=[5,-3]$, $r=3$



### ✏️ Příklad 2
Je dána rovnice:

$$
x^2 + y^2 - 2mx - 2ny + p = 0
$$

Po úpravě:

$$
(x-m)^2 + (y-n)^2 = m^2 + n^2 - p
$$

Aby šlo o kružnici, musí platit:

$$
m^2 + n^2 - p > 0
$$



### ✏️ Příklad 3 (podrobně)

Jsou dány kružnice:

$$
k_1:(x+3)^2 + (y-2)^2 = 4
$$
$$
k_2:(x-3)^2 + (y+2)^2 = 9
$$

Středy:
- $S_1=[-3,2]$
- $S_2=[3,-2]$

Vzdálenost středů:

$$
|S_1S_2| = \sqrt{(3+3)^2 + (-2-2)^2} = \sqrt{36+16} = \sqrt{52}
$$



### ✏️ Příklad 4 (vzájemná poloha)

Poloměry:
- $r_1=2$
- $r_2=3$

Součet poloměrů:
$$
r_1+r_2=5
$$

Protože:
$$
\sqrt{52} > 5
$$

👉 Kružnice **nemají žádné společné body**.



## 🔹 Příklady k procvičení

1. Urči střed a poloměr:
$$
(x+1)^2 + (y-4)^2 = 16
$$

2. Převeď na středový tvar:
$$
x^2 + y^2 - 4x + 10y + 4 = 0
$$

3. Urči, pro jaké hodnoty $p$ jde o kružnici:
$$
x^2 + y^2 + 6x + 4y + p = 0
$$

4. Rozhodni vzájemnou polohu kružnic:
$$
(x-1)^2 + y^2 = 1
$$
$$
(x-4)^2 + y^2 = 4
$$



## 🔹 Shrnutí

- kružnice je množina bodů se stejnou vzdáleností od středu
- **středový tvar**:
$$
(x-m)^2 + (y-n)^2 = r^2
$$
- **obecný tvar**:
$$
x^2 + y^2 + ax + by + c = 0
$$
- převod pomocí **doplnění na čtverec**
- vzájemná poloha kružnic závisí na vzdálenosti středů a poloměrech
