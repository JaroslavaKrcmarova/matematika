# Lineární lomená funkce

## Motivace

S lineární lomenou funkcí se setkáváme v situacích, kdy se jedna veličina mění
**nepřímo úměrně** jiné veličině, případně je tato závislost ještě posunutá.

Typickými příklady jsou:
- závislost ceny na počtu kusů,
- fyzikální vztahy (rychlost, odpor),
- ekonomické modely.

Grafem těchto funkcí je **hyperbola**.

## Definice a vlastnosti

**Lineární lomená funkce** je každá funkce daná předpisem:

$$
y = \frac{ax + b}{cx + d}
$$

kde:

* $a, b, c, d \in \mathbb{R}$,
* $c \ne 0$,
* $ad - bc \ne 0$.


### Poznámka

Pokud by platilo $c = 0$, dostali bychom lineární funkci:

$$
y = \frac{ax + b}{d} = kx + q
$$


## Asymptoty

Asymptoty lineární lomené funkce jsou přímky:

$$
x = -\frac{d}{c}
\quad \text{a} \quad
y = \frac{a}{c}
$$

Průsečík asymptot je **střed hyperboly**:

$$
S = \left[ -\frac{d}{c}, \frac{a}{c} \right]
$$

## Středový tvar funkce

Lineární lomenou funkci lze upravit do tvaru:

$$
y = y_0 + \frac{k}{x - x_0}
$$


kde platí:

$$
x_0 = -\frac{d}{c}, \quad
y_0 = \frac{a}{c}, \quad
k = \frac{bc - ad}{c^2}
$$


Znaménko výrazu $ad - bc$ určuje orientaci větví hyperboly.
